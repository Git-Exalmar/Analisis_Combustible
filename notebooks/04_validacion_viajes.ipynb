```python
# Permite importar src/ desde la carpeta notebooks/
import sys
from pathlib import Path
RAIZ = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(RAIZ))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.config import DIR_DATA_RAW

pd.set_option("display.max_columns", 50)
pd.set_option("display.float_format", "{:,.2f}".format)
sns.set_theme(style="whitegrid")
```

```python
from src.config import DIR_DATA_PROC
viajes = pd.read_parquet(DIR_DATA_PROC / "viajes.parquet")
calas = pd.read_parquet(DIR_DATA_PROC / "calas.parquet")
inval = pd.read_parquet(DIR_DATA_PROC / "viajes_invalidos.parquet")
display(pd.read_csv(DIR_DATA_PROC / "limpieza_viajes.csv"))

fig, ax = plt.subplots(2, 2, figsize=(15, 9))
viajes["ZarpeCorregido"].dt.to_period("M").value_counts().sort_index().plot.bar(ax=ax[0, 0], color="#0065AC")
ax[0, 0].set_title("Viajes por mes")
viajes["DuracionH"].clip(upper=80).plot.hist(bins=40, ax=ax[0, 1], color="#0065AC")
ax[0, 1].set_title("Duración del viaje (h)")
viajes["NCalas"].value_counts().sort_index().plot.bar(ax=ax[1, 0], color="#0065AC")
ax[1, 0].set_title("Calas por viaje (0 = sin pesca)")
viajes["PuertoZarpe"].value_counts().plot.barh(ax=ax[1, 1], color="#0065AC")
ax[1, 1].set_title("Viajes por puerto de zarpe")
plt.tight_layout()

print("Por qué se invalidaron:")
display(inval.assign(
    sin_fin=inval["FinViaje"].isna(),
    muy_largo=inval["DuracionH"] > 120,
    muy_corto=inval["DuracionH"] < 0.5,
)[["sin_fin", "muy_largo", "muy_corto"]].sum())
```
